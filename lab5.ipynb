```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
image = cv2.imread("parrot.jpeg")
image= cv2.cvtColor(img , cv2.COLOR_BGR2RGB)
# Gaussian sigma
sigma=2

#parameter
amount=1.5

# Convert image to float32 for processing
image_float = image.astype(np.float32) / 255.0

# Apply Gaussian blur
blurred = cv2.GaussianBlur(image_float, (0, 0), sigmaX=sigma, sigmaY=sigma)

# Perform Unsharp Masking
sharpened = np.clip(image_float + amount * (image_float - blurred), 0, 1)
# Plot original, blurred, and sharpened images
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")

ax[1].imshow(blurred)
ax[1].set_title("Blurred Image")
ax[1].axis("off")

ax[2].imshow(sharpened)
ax[2].set_title("Sharpened (Unsharp Masking)")
ax[2].axis("off")

plt.show() 
```

```python
import cv2
import matplotlib.pyplot as plt
from skimage import data

img = data.chelsea()

# Kernel size must be odd; sigmaX=0 lets OpenCV compute sigma from the size
gauss = cv2.GaussianBlur(img, (7, 7), 0)

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].imshow(img);   axes[0].set_title("Original")
axes[1].imshow(gauss); axes[1].set_title("Gaussian filter (7x7)")
for ax in axes:
    ax.axis("off")
plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
image_float = img.astype(np.float32) / 255.0

laplacian = cv2.Laplacian(image_float, cv2.CV_32F)
sharpened = np.clip(image_float - laplacian, 0, 1)

fig, axes = plt.subplots(1, 2, figsize=(10, 7))
axes[0].imshow(img);       axes[0].set_title("image")
axes[1].imshow(sharpened); axes[1].set_title("sharpened")
for ax in axes:
    ax.axis("off")
plt.tight_layout()
plt.show()
```
