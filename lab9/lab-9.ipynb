```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
image = "Erosion.jpg"
img = cv2.imread(image , cv2.IMREAD_GRAYSCALE)
```

```python
# elotion



_,binary = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

kirnal_1=np.ones((5,5), np.uint8)
kirnal_2=np.ones((7,7), np.uint8)
kirnal_3=np.ones((11,11), np.uint8)

eroded_1 = cv2.erode(binary,kirnal_1,iterations=1 )
eroded_2 = cv2.erode(binary, kirnal_2, iterations=1)
eroded_3 = cv2.erode(binary, kirnal_3, iterations=1)

plt.figure(figsize=(12,8))

plt.subplot(2,2,1),plt.imshow(binary),plt.title("org image")
plt.subplot(2,2,2), plt.imshow(eroded_1),plt.title("3,3 kirnal")
plt.subplot(2,2,3), plt.imshow(eroded_2),plt.title("7,7 kirnal")
plt.subplot(2,2,4), plt.imshow(eroded_3), plt.title("11,11 kirnal")

plt.show()

```

```python
# delation 
dire_1 = cv2.dilate(binary , kirnal_1, iterations = 1)
dire_2 = cv2.dilate(binary , kirnal_2, iterations = 1)
dire_3 = cv2.dilate(binary , kirnal_3, iterations = 1)

plt.figure(figsize=(12,8))

plt.subplot(2,2,1),plt.imshow(binary),plt.title("org image")
plt.subplot(2,2,2), plt.imshow(dire_1), plt.title("3,3 kirnal")
plt.subplot(2,2,3), plt.imshow(dire_2), plt.title("7,7 kirnal")
plt.subplot(2,2,4), plt.imshow(dire_3), plt.title("11,11 kirnal")

```

```python
# opening 

image = "Open_Closing.jpg"

img = cv2.imread(image , cv2.IMREAD_GRAYSCALE)

_,binary = cv2.threshold(img,0,255,cv2.THRESH_BINARY + cv2.THRESH_OTSU)

kernel = np.ones((11,11), np.uint8)

opening = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)

plt.figure(figsize=(12,8))

plt.subplot(1,2,1), plt.imshow(binary), plt.title("org image")
plt.subplot(1,2,2), plt.imshow(opening), plt.title("opening")

plt.show()
```

```python
#close
closing = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)
plt.figure(figsize=(12,8))

plt.subplot(2,2,1), plt.imshow(binary), plt.title("org image")
plt.subplot(2,2,2), plt.imshow(closing), plt.title("closing")
```

```python
image = "Balls(1).png"
img = cv2.imread(image,0)

_,binary = cv2.threshold(img,0,255,cv2.THRESH_BINARY + cv2.THRESH_OTSU)
kernel_1=np.ones((3,3), np.uint8)
kernel_2=np.ones((7,7), np.uint8)
kernel_3=np.ones((11,11), np.uint8)

#erosion

erosion_1 = cv2.erode(binary, kernel_1, iterations=1)
erosion_2 = cv2.erode(binary, kernel_2, iterations=1)
erosion_3 = cv2.erode(binary, kernel_3, iterations=1)

# dilation
dilation_1 = cv2.dilate(binary, kernel_1, iterations=1)
dilation_2 = cv2.dilate(binary, kernel_2, iterations=1)
dilation_3 = cv2.dilate(binary, kernel_3, iterations=1)

plt.figure(figsize=(12,8))

plt.subplot(2,2,1), plt.imshow(binary), plt.title("org image")
plt.subplot(2,2,2), plt.imshow(erosion_1), plt.title("erosion 3,3 kernel")
plt.subplot(2,2,3), plt.imshow(erosion_2), plt.title("erosion 7,7 kernel")
plt.subplot(2,2,4), plt.imshow(erosion_3), plt.title("erosion 11,11 kernel")
```
